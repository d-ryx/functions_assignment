Write a function to solve a second order differential equation.let the differential equation have  both homogeneous solution and particular solution as output ask the user to input auxiliary conditions

```python
import sympy as sp

def solve_ode():
    x = sp.Symbol('x')
    y = sp.Function('y')

    a = float(input("Enter coefficient of y'': "))
    b = float(input("Enter coefficient of y': "))
    c = float(input("Enter coefficient of y: "))
    f_x = sp.sympify(input("Enter f(x): "))

    ode = sp.Eq(a*y(x).diff(x, 2) + b*y(x).diff(x) + c*y(x), f_x)

    # Find Homogeneous Solution (y_h)
    y_h = sp.dsolve(ode.subs(f_x, 0), y(x)).rhs

    # Find the Particular Solution (y_p)
    y_p = sp.dsolve(ode, y(x)).rhs - y_h

    print(f"\nHomogeneous Solution (y_h): {y_h}")
    print(f"Particular Solution (y_p): {y_p}")

    x0 = float(input("Enter x0: "))
    y0 = float(input(f"Enter y({x0}): "))
    v0 = float(input(f"Enter y'({x0}): "))

    conditions = {y(x).subs(x, x0): y0, y(x).diff(x).subs(x, x0): v0}

    final_sol = sp.dsolve(ode, y(x), ics=conditions)

    print(f"\nThe solution is: {final_sol.rhs}")
```

```python
s2_ode = solve_ode()
```
